```python
import pandas as pd
```

```python
#проверочка данных на нормальность
accounts = pd.read_csv('ravenstack_accounts.csv')
churn = pd.read_csv('ravenstack_churn_events.csv')
feature = pd.read_csv('ravenstack_feature_usage.csv')
```

```python
print(accounts.head())
print(churn.head())
```

```python
#Надо подсчитать Churn Rate, то есть процент возврата пользователей
total_accounts = accounts['account_id'].nunique()
churned_accounts = churn['account_id'].nunique()
churn_rate = (churned_accounts / total_accounts) * 100
```

```python
print(f"Всего клиентов в базе: {total_accounts}")
print(f"Из них ушло (отток): {churned_accounts}")
print(f"Общий Churn Rate компании: {churn_rate:.2f}%")
```

```python
#время визуализации для поиска корня проблемы
#а проблема есть, тк Churn Rate в 70% - это катастрофа
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")
```

```python
sns.countplot(data=accounts, x='plan_tier', hue='churn_flag', palette='Set2')

plt.title('Распределение оттока клиентов по тарифным планам', fontsize=14, pad=15)
plt.xlabel('Тарифный план (Plan Tier)', fontsize=12)
plt.ylabel('Количество компаний', fontsize=12)
plt.legend(title='Ушел клиент? (Churn)', labels=['Нет (False)', 'Да (True)'])

plt.show()
```
